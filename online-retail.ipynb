```python
df = df.fillna({'Customer ID': 'Guest', 'Description': 'Unknown Product'})
```

```python
print(df.isna().sum())
```

```python
print(f"Jumlah baris sebelum dibersihkan: {len(df)}")

df = df[df['Price'] > 0]

kondisi_retur_resmi = df['Invoice'].astype(str).str.startswith('C')
kondisi_kuantitas_minus = df['Quantity'] < 0

df = df[(df['Quantity'] > 0) | (kondisi_kuantitas_minus & kondisi_retur_resmi)]
print(f"Jumlah baris setelah dibersihkan: {len(df)}")
```

```python
df.to_csv('/content/drive/MyDrive/Onlien Retail Analyst/online_retail_cleaned.csv', index=False)
print("Data yang sudah di cleaning berhasil disimpan")
```
